```python
import datetime

# Asks the user for the event title, year, month, and day.
event_title = input("Please enter the event title: ")
year = int(input("Please input the year: "))
month = int(input("Please enter month: "))
day = int(input("Please enter day: "))

full_date = datetime.date(year, month, day)

# Formats the inputs.
format_event = event_title.strip().upper()
format_date = full_date.strftime("%B %d %Y")
format_month = full_date.strftime("%B")
weekday = full_date.strftime("%A")

print("-----------------")
print("FORMATTED OUTPUT")
print("Event:", format_event)
print("Full Date:", format_date)
print("Month:", format_month)
print("Weekday:", weekday)
```
